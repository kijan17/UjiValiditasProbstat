```python
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from google.colab import files

print("Langkah 1 & 2: Mengunggah Berkas Data")
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
# Membaca seluruh data dari Excel
df_mentah = pd.read_excel(nama_file)

# Memfilter HANYA responden dari kelas 3 TI A
df = df_mentah[df_mentah['Kelas'].astype(str).str.contains('3 TI A', case=False, na=False)].copy()

# Otomatis mengambil kolom yang mengandung tanda kurung siku [...]
kolom_kuesioner = [kol for kol in df.columns if '[' in kol and ']' in kol]
butir = df[kolom_kuesioner]

print("Jumlah responden 3 TI A:", len(df))
print("Jumlah butir pertanyaan:", butir.shape[1])
print("Data kosong:", int(butir.isna().sum().sum()))
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
    # Antisipasi format teks tanpa singkatan
    "Sangat Tidak Setuju": 1,
    "Tidak Setuju": 2,
    "Netral": 3,
    "Setuju": 4,
    "Sangat Setuju": 5
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
    # Antisipasi format teks tanpa singkatan
    "Sangat Tidak Setuju": 1,
    "Tidak Setuju": 2,
    "Netral": 3,
    "Setuju": 4,
    "Sangat Setuju": 5
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
dimensi = {}
for kolom in butir.columns:
    cocok = re.match(r"^(.*?)\s*\[", kolom)
    if cocok:
        nama = cocok.group(1).strip()
    else:
        nama = "Dimensi Umum"
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")
```

```python
n = len(num)
t = stats.t.ppf(0.975, n-2)
r_tabel = t / np.sqrt((n-2) + t**2)
print(f"n = {n} | r tabel (alpha=0,05; df={n-2}) = {r_tabel:.3f}\n")

hasil_validitas = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)
    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid"
        })

tabel_validitas = pd.DataFrame(hasil_validitas)
display(tabel_validitas)
```

```python
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
    return ("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
            else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)
    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a)
    })
    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(cronbach_alpha(x.drop(columns=kolom)), 3)
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)

display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
stat_deskriptif = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    skor_rata = x.mean(axis=1)

    # Menghitung Standar Deviasi, Skewness, Kurtosis
    stat_deskriptif.append({
        "Dimensi": nama,
        "Mean": round(skor_rata.mean(), 3),
        "Standar Deviasi": round(skor_rata.std(ddof=1), 3),
        "Skewness": round(stats.skew(skor_rata, bias=False), 3),
        "Kurtosis": round(stats.kurtosis(skor_rata, bias=False), 3)
    })

    # Membuat Grafik Komposisi Jawaban per Dimensi
    rekap_jawaban = pd.DataFrame()
    for i, kol in enumerate(daftar, 1):
        hitung = x[kol].value_counts().reindex([1, 2, 3, 4, 5], fill_value=0)
        rekap_jawaban[f"B{i}"] = hitung

    rekap_jawaban.index = ["1(STS)", "2(TS)", "3(N)", "4(S)", "5(SS)"]
    rekap_jawaban.T.plot(kind="bar", stacked=True, figsize=(8, 4), colormap="RdYlGn", edgecolor="black")
    plt.title(f"Komposisi Jawaban - {nama}")
    plt.legend(bbox_to_anchor=(1.02, 1))
    plt.tight_layout()
    plt.show()

tabel_deskriptif = pd.DataFrame(stat_deskriptif)
display(tabel_deskriptif)
```

```python
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")

with pd.ExcelWriter("Hasil_Validitas_Reliabilitas_3TIA.xlsx") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)
    tabel_deskriptif.to_excel(w, sheet_name="Deskriptif", index=False)

files.download("Hasil_Validitas_Reliabilitas_3TIA.xlsx")
print("File Excel berhasil diunduh!")
```
